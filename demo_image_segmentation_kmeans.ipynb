```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
import glob
from functools import partial

from sklearn.cluster import KMeans

# from JSAnimation import IPython_display
from matplotlib import animation
from IPython.display import HTML
from tqdm.notebook import tqdm
from matplotlib.colors import ListedColormap

import matplotlib as mpl
mpl.rcParams['animation.embed_limit'] = 2**128
```

```python
portraits = glob.glob("images/demos/pictures/*.jpg") + glob.glob("images/demos/pictures/*.png")
portraits
```

```python
def load_image(path):
    im_frame = Image.open(path)
    np_frame = np.array(im_frame.get_flattened_data())
    width, height = im_frame.size
    print(np_frame.shape)
    if path.endswith(".png"):
        x = np_frame.reshape(height * width, 4)
    else:
        x = np_frame.reshape(height * width, 3)
    return x[:, :3], width, height
```

```python
def plot_face(port, ax=None, title=None):
    global width, height 
    if ax is None:
        fig, ax = plt.subplots(figsize=(6, 6))
    ax.imshow(port.reshape(height, width, 3).astype('uint8'))
    ax.axis('off')
    if title:
        title = ax.set_title(title)
```

```python
def animateSegmentation(i, x=None, progress_callback=None):
    r = i + 1
    kmeans = KMeans(n_clusters=r)
    kmeans.fit(x)
    Y = np.zeros([x.shape[0], r])
    Y[np.arange(Y.shape[0]), kmeans.labels_] = 1
    centroids = kmeans.cluster_centers_/256
    # unique, counts = np.unique(kmeans.labels_, return_counts=True)    

    cmap = ListedColormap(centroids)
    color_map_ax.imshow([np.arange(r)], cmap=cmap, aspect='auto')
    color_map_ax.axis('off')
    
    ax.cla()
    plot_face(Y @ kmeans.cluster_centers_, ax=ax, title=f"r={r}")

    if progress_callback:
        progress_callback()
```

```python
fig = plt.figure(layout="constrained", figsize=(12, 6),)
subfigs = fig.subfigures(2, 1,  wspace=0.07, height_ratios=[8, 1.])
left_ax, ax = subfigs[0].subplots(1, 2)
color_map_ax = subfigs[1].subplots(1, 1)

x, width, height = load_image('images/demos/portraits/afeghan_girl.jpg')
plot_face(x, ax=left_ax, title="Afghan girl (Steve Mccurry, 1984)")
number_of_frames = 30
with tqdm(unit='frames', total=number_of_frames, leave=False) as pbar:
    anim = animation.FuncAnimation(fig, partial(animateSegmentation,x = x, progress_callback=pbar.update), frames=number_of_frames, interval=600, blit=False)
    html = anim.to_jshtml(default_mode='once')
plt.close()
HTML(html)
```
