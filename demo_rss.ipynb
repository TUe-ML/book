```python
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np

from matplotlib.patches import Rectangle
from matplotlib.axes import Axes

import ipywidgets as widgets
```

```python
def fit_line(x, y):
    """Return the fitted coefficients, predictions, and RSS."""
    coef = np.polyfit(x, y, 1)
    y_pred = np.polyval(coef, x)
    rss = np.sum((y - y_pred) ** 2)
    return coef, y_pred, rss


def setup_plot(ax: Axes):
    ax.set_ylim(-7, 6)
    ax.set_aspect("equal", adjustable="box")


def plot_points(ax: Axes, x, y, label_indices=(), labels=()):
    ax.scatter(x, y, color="blue", s=60)

    for label_index, label in zip(label_indices, labels):
        ax.annotate(
            label,
            xy=(x[label_index], y[label_index]),
            xytext=(15, 15),
            textcoords="offset points",
            arrowprops={"arrowstyle": "->"},
            fontsize=14,
        )


def plot_fitted_line(ax, x, coef):
    xx = np.linspace(x.min(), x.max(), 200)
    ax.plot(xx, np.polyval(coef, xx), color="black", lw=2)


def plot_residual_squares(ax, x, y, y_pred):
    for xi, yi, yhat in zip(x, y, y_pred):
        residual = yi - yhat
        side = abs(residual)
        ax.add_patch(
            Rectangle(
                (xi, min(yi, yhat)),
                side,
                side,
                facecolor="red",
                edgecolor="red",
                alpha=0.35,
            )
        )


def plot_rss_value(ax, rss):
    legend_patch = Rectangle(
        (0, 0),
        1,
        1,
        facecolor="red",
        edgecolor="red",
        alpha=0.35,
        label=f"RSS: {rss:.1f}",
    )
    ax.legend(handles=[legend_patch], loc="lower right", fontsize=14)



def plot_rss(
    ax,
    x,
    y,
    label_indices=(),
    labels=(),
):
    """Plot all four RSS stages in sequence on an existing axis."""
    coef, y_pred, rss = fit_line(x, y)
    setup_plot(ax)
    plot_points(ax, x, y, label_indices, labels)
    # save_snapshot(fig, filename_generator)

    plot_fitted_line(ax, x, coef)
    # save_snapshot(fig, filename_generator)
    plot_residual_squares(ax, x, y, y_pred)
    plot_rss_value(ax, rss)
    # save_snapshot(fig, filename_generator)
```

```python
x2 = np.array([-5, -3.8, -2.8, -1.7, -0.5, 0.6, 1.6, 2.8, 4.0, 5.0, -4.0])

fig, ax = plt.subplots(figsize=(6, 6))
plt.close(fig)
def update(y):
    ax.clear()
    ax.set_axis_off()
    y2 = np.array(
        [-6.1, -2.9, -2.5, -3.2, -1.1, 2.2, -0.8, 2.3, 5.1, 4.1, float(y)]
    )
    plot_rss(
        ax,
        x2,
        y2,
        label_indices=[-1],
        labels=["outlier"],
    )
    fig.canvas.draw_idle()
    return fig


widgets.interact(
    update,
    y=widgets.FloatSlider(
        value=-5,
        min=-5,
        max=5,
        step=0.25,
        description="y of outlier:",
        continuous_update=True,
    ),
);
```
