# Regression demo

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
D = np.array([1,2,3])
y = np.array([1,2,2])
```

```python
def ϕ_aff(D):
    return np.column_stack((np.ones(D.shape[0]),D))
```

```python
X=ϕ_aff(D)
```

```python
X
```

```python
y
```

```python
X.T@X
```

```python
X.T@y
```

```python
np.linalg.solve(X.T@X,X.T@y)
```

```python
np.linalg.inv(X.T@X)@X.T@y
```

```python
plt.figure(figsize=(14, 5))
x = np.linspace(0, 6, 100)
β = np.linalg.inv(X.T@X)@X.T@y
f_x = ϕ_aff(x)@β
plt.plot(x, f_x, label="$f_\\beta$")
plt.scatter(D, y, edgecolor='b', s=50)
plt.xlabel("x")
plt.ylabel("y")
plt.xlim((0.5, 3.5))
plt.legend(loc="best")
plt.show()
```
